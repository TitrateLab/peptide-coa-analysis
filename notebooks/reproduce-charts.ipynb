# Reproduce every chart from data/latest

This notebook renders the charts in `charts/` from the release files only. It needs `matplotlib` and `jinja2` (`pip install -r requirements.txt`) and nothing else. No network access, no private data.


```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if not (ROOT / 'scripts' / 'build_release.py').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'scripts'))
import build_release as br

problems = br.check(ROOT / 'data' / 'latest')
assert not problems, problems
data = br.load(ROOT / 'data' / 'latest')
view = br.build_view(data)
print('release', view['m']['version'], 'included records', view['included'])
```

```python
# Renders light and dark SVGs plus a PNG for each chart into charts/.
br.render_charts(view)
sorted(p.name for p in (ROOT / 'charts').iterdir())
```

```python
# The table behind the first chart: share of tested vials within 10% of the printed label.
for row in view['q_chart']:
    print(f"{row['compound']:<32} {row['quantity_results']:>7,}  {row['quantity_within_10pct_pct']:>5.1f}%")
```

```python
from IPython.display import SVG, display
for name in ['quantity-within-label', 'endotoxin-testing-rate', 'zero-identity-share', 'testing-volume', 'compound-volume', 'lab-mix']:
    display(SVG(filename=str(ROOT / 'charts' / f'{name}-light.svg')))
```
